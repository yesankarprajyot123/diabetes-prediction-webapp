```python
df[['Glucose',
    'BloodPressure',
    'SkinThickness',
    'Insulin',
    'BMI']].eq(0).sum()
```

```python
import numpy as np

cols = [
    'Glucose',
    'BloodPressure',
    'SkinThickness',
    'Insulin',
    'BMI'
]

df[cols] = df[cols].replace(0, np.nan)
```

```python
df.isnull().sum()
```

```python
for col in cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.isnull().sum()
```
